```python
def update_savings(current_total, deposit):
    updated_total = current_total + deposit
    return updated_total


def check_goal(total, target):
    if total >= target:
        return "Goal reached!"
    else:
        return "Keep saving!"


target_input = input("Enter your savings target: ")
savings_target = float(target_input)

running_total = 0.0
current_deposit = -1.0

while current_deposit != 0:
    deposit_input = input("Enter deposit amount: ")
    current_deposit = float(deposit_input)

    if current_deposit != 0:
        running_total = update_savings(running_total, current_deposit)

print(f"Total Savings: {running_total:.2f}")

status = check_goal(running_total, savings_target)
print(f"Status: {status}")

if running_total < savings_target:
    remaining_needed = savings_target - running_total
    print(f"Remaining Amount Needed: {remaining_needed:.2f}")
```
