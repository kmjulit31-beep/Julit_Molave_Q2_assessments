```python
def add_numbers(num1, num2):                # function for addition
    return num1 + num2                      # add and return

def subtract_numbers(num1, num2):           # function for subtraction
    return num1 - num2                      # subtract and return

def multiply_numbers(num1, num2):           # function for multiplication
    return num1 * num2                      # multiply and return

def divide_numbers(num1, num2):             # function for division
    if num2 == 0:                           # check if zero
        return "Cannot divide by zero"      # error if zero
    return num1 / num2                      # divide and return

num1 = float(input("Enter first number: "))      # ask first number
num2 = float(input("Enter second number: "))     # ask second number

print("1 - Addition")                       # show 1
print("2 - Subtraction")                    # show 2
print("3 - Multiplication")                 # show 3
print("4 - Division")                       # show 4
choice = input("Enter choice: ")            # ask choice

if choice == "1":                           # if choice 1
    result = add_numbers(num1, num2)        # call add

if choice == "2":                           # if choice 2
    result = subtract_numbers(num1, num2)   # call subtract

if choice == "3":                           # if choice 3
    result = multiply_numbers(num1, num2)   # call multiply

if choice == "4":                           # if choice 4
    result = divide_numbers(num1, num2)     # call divide

print("Result:", result)                    # show result

# 1. What functions did you create?
# add_numbers, subtract_numbers, multiply_numbers, divide_numbers
#
# 2. What parameters did your functions use?
# num1 and num2
#
# 3. What arguments were passed when the functions were called?
# the numbers the user entered
#
# 4. How did your program use the returned value?
# stored in result and printed
#
# 5. Why is it better to divide the program into functions rather than writing everything in one long block of code?
# because it makes the code cleaner, organized, and easier to fix
```
