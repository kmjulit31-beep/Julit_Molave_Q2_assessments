```python
scores = [85, 90, 78, 92, 88]

def display_scores():
    print("\n--- Student Scores ---")
    for i in range(len(scores)):
        print("Index", i, ":", scores[i])

def search_score():
    target = int(input("\nEnter score to search: "))
    if target in scores:
        print("Score found at index:", scores.index(target))
    else:
        print("Score not found.")

def compute_average():
    average = sum(scores) / len(scores)
    print("\nAverage Score:", average)

def update_score():
    display_scores()
    index = int(input("Enter index to update (0 to 4): "))
    new_score = int(input("Enter new score: "))
    scores[index] = new_score
    print("Score updated!")

while True:
    print("\n=== Student Score Analyzer ===")
    print("Select the number of the operation you want to perform:")
    print("1. Display Scores")
    print("2. Search Score")
    print("3. Compute Average Score")
    print("4. Update Score")
    print("5. Display Updated Scores")
    print("6. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":
        display_scores()
    elif choice == "2":
        search_score()
    elif choice == "3":
        compute_average()
    elif choice == "4":
        update_score()
    elif choice == "5":
        display_scores()
    elif choice == "6":
        print("Goodbye!")
        break
    else:
        print("Invalid choice. Try again.")
```
